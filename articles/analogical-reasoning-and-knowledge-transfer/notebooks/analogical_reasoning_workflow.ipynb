# Analogical Reasoning and Knowledge Transfer

This notebook scaffold loads the analogical reasoning dataset, computes descriptive summaries, and prepares data for modeling mapping accuracy, transfer success, inference quality, schema abstraction, relational complexity, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/analogical_reasoning_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/analogical_reasoning_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('mapping_accuracy', 'size'),
        mapping_rate=('mapping_accuracy', 'mean'),
        transfer_rate=('transfer_success', 'mean'),
        mean_structural_similarity=('structural_similarity', 'mean'),
        mean_surface_similarity=('surface_similarity', 'mean'),
        mean_inference_quality=('inference_quality', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
