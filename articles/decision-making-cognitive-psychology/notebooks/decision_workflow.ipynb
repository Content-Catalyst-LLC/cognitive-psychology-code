# Decision Making in Cognitive Psychology

This notebook scaffold loads the decision-making dataset and prepares summaries for risky choice, expected value, subjective value, framing, evidence accumulation, confidence, response time, AI agreement, verification burden, and decision quality.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/decision_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/decision_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('choice_risky', 'size'),
        risky_choice_rate=('choice_risky', 'mean'),
        optimal_choice_rate=('optimal_choice', 'mean'),
        mean_accuracy=('accuracy', 'mean'),
        mean_confidence=('confidence', 'mean'),
        mean_decision_quality=('decision_quality', 'mean'),
        mean_response_time_ms=('response_time_ms', 'mean'),
        mean_uncertainty=('uncertainty', 'mean'),
        mean_verification_burden=('verification_burden', 'mean')
    )
    .reset_index()
)
summary
```
