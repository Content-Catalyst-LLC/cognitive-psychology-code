# Mental Models in Cognitive Psychology

This notebook scaffold loads the mental-model dataset and prepares summaries for model completeness, coherence, causal-link accuracy, feedback-loop recognition, prediction error, problem success, model revision, transfer, and reasoning time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/mental_models_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/mental_models_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('problem_success', 'size'),
        mean_prediction_error=('prediction_error', 'mean'),
        mean_understanding=('system_understanding_score', 'mean'),
        success_rate=('problem_success', 'mean'),
        intervention_accuracy=('intervention_choice_accuracy', 'mean'),
        mean_transfer=('transfer_score', 'mean'),
        mean_cognitive_load=('cognitive_load', 'mean')
    )
    .reset_index()
)
summary
```
