# Cognitive Psychology and Behavioral Economics

This notebook scaffold loads the behavioral-economics dataset, computes descriptive summaries, and prepares data for modeling risky choice, default acceptance, willingness to pay, cognitive load, framing, delay, and decision time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/behavioral_economics_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/behavioral_economics_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('risky_choice', 'size'),
        risky_choice_rate=('risky_choice', 'mean'),
        default_acceptance_rate=('default_accepted', 'mean'),
        mean_cognitive_load=('cognitive_load', 'mean'),
        mean_attention=('attention_score', 'mean'),
        mean_wtp=('willingness_to_pay', 'mean'),
        mean_decision_time=('decision_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
