# Concept Formation in Cognitive Psychology

This notebook scaffold loads the concept-formation dataset, computes descriptive summaries, and prepares data for modeling categorization accuracy, prototype distance, exemplar similarity, boundary ambiguity, generalization, discrimination, abstraction quality, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/concept_formation_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/concept_formation_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('category_accuracy', 'size'),
        accuracy_rate=('category_accuracy', 'mean'),
        mean_prototype_distance=('prototype_distance', 'mean'),
        mean_exemplar_similarity=('exemplar_similarity', 'mean'),
        mean_generalization=('generalization_score', 'mean'),
        mean_discrimination=('discrimination_score', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
