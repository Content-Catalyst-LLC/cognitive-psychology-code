# Attention in Cognitive Psychology

This notebook scaffold loads the attention dataset and prepares summaries for cueing, vigilance, signal detection, response time, attentional load, lapse probability, executive control, interface salience, and divided-attention cost.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/attention_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/attention_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('correct', 'size'),
        correct_rate=('correct', 'mean'),
        yes_rate=('response_yes', 'mean'),
        mean_rt=('rt', 'mean'),
        mean_confidence=('confidence', 'mean'),
        mean_lapse_probability=('lapse_probability', 'mean'),
        mean_vigilance_state=('vigilance_state', 'mean'),
        mean_divided_attention_cost=('divided_attention_cost', 'mean')
    )
    .reset_index()
)
summary
```
