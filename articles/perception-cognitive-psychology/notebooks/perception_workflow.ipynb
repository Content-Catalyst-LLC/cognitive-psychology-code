# Perception in Cognitive Psychology

This notebook scaffold loads the perception dataset and prepares summaries for psychometric curves, signal detection, attention gain, prediction error, confidence, response time, multisensory congruence, visual search, and interface salience.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/perception_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/perception_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('correct', 'size'),
        correct_rate=('correct', 'mean'),
        yes_rate=('response_yes', 'mean'),
        mean_confidence=('confidence', 'mean'),
        mean_response_time_ms=('response_time_ms', 'mean'),
        mean_sensory_evidence=('sensory_evidence', 'mean'),
        mean_prediction_error=('prediction_error', 'mean'),
        mean_threshold=('perceptual_threshold', 'mean')
    )
    .reset_index()
)
summary
```
