# Cognitive Constraints in Organizational Decision Making

This notebook scaffold loads the organizational decision dataset, computes descriptive summaries, and prepares data for modeling cognitive burden, decision quality, satisficing, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/organizational_decision_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/cognitive_constraints_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('decision_quality', 'size'),
        mean_burden=('cognitive_burden', 'mean'),
        mean_quality=('decision_quality', 'mean'),
        satisficing_rate=('chose_satisficing', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
