# Sensory Memory in Cognitive Psychology

This notebook scaffold loads the sensory-memory dataset and prepares summaries for modality, cue delay, trace strength, report accuracy, partial report, masking, working-memory transfer, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/sensory_memory_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/sensory_memory_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby(['modality', 'cue_delay_ms'])
    .agg(
        n_trials=('correct', 'size'),
        accuracy=('correct', 'mean'),
        mean_report=('report_score', 'mean'),
        mean_trace=('trace_strength', 'mean'),
        mean_selection=('selection_probability', 'mean'),
        wm_transfer_rate=('wm_transfer', 'mean')
    )
    .reset_index()
)
summary.head()
```
