# Cognitive Load and Information Processing

This notebook scaffold loads the cognitive-load dataset and prepares summaries for intrinsic load, extraneous load, germane processing, prior knowledge, performance, transfer, mental effort, and efficiency.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/cognitive_load_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/cognitive_load_model.py --simulate first')

df = pd.read_csv(data_path)
df['total_load'] = df['intrinsic_load'] + df['extraneous_load'] + df['germane_load']
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('correct', 'size'),
        mean_intrinsic=('intrinsic_load', 'mean'),
        mean_extraneous=('extraneous_load', 'mean'),
        mean_germane=('germane_load', 'mean'),
        mean_effort=('subjective_effort', 'mean'),
        accuracy=('performance_accuracy', 'mean'),
        correct_rate=('correct', 'mean'),
        mean_transfer=('transfer_score', 'mean'),
        mean_efficiency=('mental_efficiency', 'mean')
    )
    .reset_index()
)
summary
```
