# Semantic Memory in Cognitive Psychology

This notebook scaffold loads the semantic-memory dataset, computes descriptive summaries, and prepares data for modeling semantic verification accuracy, retrieval latency, category strength, false semantic association, and semantic-distance effects.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/semantic_memory_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/semantic_memory_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('verification_accuracy', 'size'),
        accuracy_rate=('verification_accuracy', 'mean'),
        mean_semantic_distance=('semantic_distance', 'mean'),
        mean_typicality=('category_typicality', 'mean'),
        mean_category_strength=('category_strength', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
