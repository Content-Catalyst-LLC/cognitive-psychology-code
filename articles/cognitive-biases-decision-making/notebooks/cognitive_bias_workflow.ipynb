# Cognitive Biases in Decision Making

This notebook scaffold loads the cognitive-bias dataset and prepares summaries for framing, anchoring, confirmation bias, overconfidence, calibration, prospect-theory features, debiasing, decision quality, and review flags.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/cognitive_bias_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/cognitive_bias_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('bias_type')
    .agg(
        n_trials=('correct', 'size'),
        mean_confidence=('confidence_rating', 'mean'),
        mean_accuracy=('actual_accuracy', 'mean'),
        mean_calibration_error=('calibration_error', 'mean'),
        mean_overconfidence=('overconfidence', 'mean'),
        risky_choice_rate=('chose_risky', 'mean'),
        correct_rate=('correct', 'mean'),
        mean_decision_quality=('decision_quality', 'mean'),
        review_flag_rate=('institutional_review_flag', 'mean')
    )
    .reset_index()
)
summary
```
