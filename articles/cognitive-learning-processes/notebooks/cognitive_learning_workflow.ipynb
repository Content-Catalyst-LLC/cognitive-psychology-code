# Cognitive Learning Processes

This notebook scaffold loads the cognitive-learning dataset and prepares summaries for attention, encoding, schema strength, retrieval practice, cognitive load, transfer, retention, and adaptive application.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/cognitive_learning_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/cognitive_learning_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
learning_curve = (
    df.groupby(['condition', 'session'])
    .agg(
        mean_accuracy=('accuracy', 'mean'),
        mean_comprehension=('comprehension_score', 'mean'),
        mean_transfer=('transfer_score', 'mean'),
        mean_retention=('retention_score', 'mean'),
        mean_cognitive_load=('cognitive_load', 'mean')
    )
    .reset_index()
)
learning_curve.head()
```
