# Cognitive Systems in Artificial Intelligence

This notebook scaffold loads the cognitive-systems dataset, computes descriptive summaries, and prepares data for modeling prediction accuracy, action success, explanation quality, calibration, trust, and override behavior.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/cognitive_systems_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/cognitive_systems_ai_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('architecture')
    .agg(
        n_trials=('prediction_accuracy', 'size'),
        mean_prediction_accuracy=('prediction_accuracy', 'mean'),
        action_success_rate=('action_success', 'mean'),
        mean_explanation=('explanation_score', 'mean'),
        override_rate=('override_decision', 'mean'),
        mean_calibration_error=('calibration_error', 'mean')
    )
    .reset_index()
)
summary
```
