# Metacognition: Thinking About Thinking

This notebook scaffold loads the metacognition dataset, computes calibration summaries, and prepares data for modeling confidence, accuracy, monitoring, control, review choice, strategy shift, feedback use, study-time allocation, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/metacognition_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/metacognition_model.py --simulate first')

df = pd.read_csv(data_path)
df['calibration_error'] = (df['confidence_rating'] - df['actual_accuracy']).abs()
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('actual_accuracy', 'size'),
        accuracy_rate=('actual_accuracy', 'mean'),
        mean_confidence=('confidence_rating', 'mean'),
        mean_calibration_error=('calibration_error', 'mean'),
        strategy_shift_rate=('strategy_shift', 'mean'),
        review_choice_rate=('review_choice', 'mean'),
        mean_regulation=('metacognitive_regulation_score', 'mean')
    )
    .reset_index()
)
summary
```
