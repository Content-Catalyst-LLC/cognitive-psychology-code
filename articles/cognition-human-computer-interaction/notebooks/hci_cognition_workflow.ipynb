# Cognition in Human-Computer Interaction

This notebook scaffold loads the HCI cognition dataset, computes descriptive summaries, and prepares data for modeling task success, cognitive load, response time, errors, warning detection, mental-model alignment, and interface condition effects.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/hci_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/hci_cognition_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('interface_condition')
    .agg(
        n_trials=('success', 'size'),
        success_rate=('success', 'mean'),
        mean_cognitive_load=('cognitive_load', 'mean'),
        mean_alignment=('alignment_score', 'mean'),
        mean_errors=('error_count', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
