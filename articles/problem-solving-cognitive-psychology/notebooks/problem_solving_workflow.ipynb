# Problem Solving in Cognitive Psychology

This notebook scaffold loads the problem-solving dataset, computes descriptive summaries, and prepares data for modeling solution accuracy, solution quality, working-memory load, strategy selection, strategy switching, insight events, and response time.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/problem_solving_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/problem_solving_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('solution_accuracy', 'size'),
        accuracy_rate=('solution_accuracy', 'mean'),
        mean_quality=('solution_quality', 'mean'),
        mean_representation=('representation_quality', 'mean'),
        mean_wm_load=('wm_load', 'mean'),
        switch_rate=('switched_strategy', 'mean'),
        mean_response_time=('response_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
