# Language Processing in Cognitive Psychology

This notebook scaffold loads the language-processing dataset, computes descriptive summaries, and prepares data for modeling comprehension accuracy, lexical decision, production accuracy, reading time, working-memory load, syntactic complexity, semantic predictability, pragmatic inference, and discourse coherence.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/language_processing_trials.csv')
if not data_path.exists():
    raise FileNotFoundError('Run python/language_processing_model.py --simulate first')

df = pd.read_csv(data_path)
df.head()
```

```python
summary = (
    df.groupby('condition')
    .agg(
        n_trials=('comprehension_accuracy', 'size'),
        comprehension_accuracy=('comprehension_accuracy', 'mean'),
        lexical_accuracy=('lexical_decision_accuracy', 'mean'),
        production_accuracy=('production_accuracy', 'mean'),
        mean_syntax=('syntactic_complexity', 'mean'),
        mean_wm_load=('working_memory_load', 'mean'),
        mean_reading_time=('reading_time_ms', 'mean')
    )
    .reset_index()
)
summary
```
